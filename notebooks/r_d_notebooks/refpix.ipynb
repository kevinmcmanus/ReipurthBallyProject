```python
def wcs_adjust1(wcsvals, rootdir, caldir='calibrated', adjdir='wcs_adjust'):
    outdir = os.path.join(rootdir, adjdir+'1')
    indir = os.path.join(rootdir, caldir)

    if not os.path.exists(outdir):
        os.mkdir(outdir)

    for f in wcsvals.iterrows():
        r=f[1]
        fin = os.path.join(indir,r.frameID+'.fits')
        with fits.open(fin) as hdul:
            h = hdul[0].header.copy()
            w = WCS(h)

            #update this wcs with the Gaia data
            w.wcs.crpix = [r.CRPIX1, r.CRPIX2]
            w.wcs.crval = [r.CRVAL1, r.CRVAL2]

            #where is ZCMa in the updated wcs?
            ZCMa_pix = w.world_to_pixel(ZCMa_coord)
            h['CRPIX1'] = int(ZCMa_pix[0])+1 #for fits index convention
            h['CRPIX2'] = int(ZCMa_pix[1])+1 #for fits index convention
            h['CRVAL1'] = ZCMa_coord.ra.value
            h['CRVAL2'] = ZCMa_coord.dec.value

            phdu = fits.PrimaryHDU(data = hdul[0].data,
                                    header=h)
            fout = os.path.join(outdir, r.frameID+'.fits')
            phdu.writeto(fout, overwrite=True)
```

```python
# see if objects in different frames are predicted by delta CRPIX
def predpix(wcsvals, imdir):
    #reference frame is first one in wcsvals
    refframe = wcsvals.iloc[0]
    refpath = os.path.join(imdir, refframe.frameID+'.fits')
    with fits.open(refpath) as hdul:
        refhdr = hdul[0].header
        refval = np.array([refhdr['CRVAL1'], refhdr['CRVAL2']])
        refdelta = np.array([refhdr['CDELT1'], refhdr['CDELT2']])
    
    refpix = np.array([refframe.CRPIX1, refframe.CRPIX2])
    print(f'Reference Pixels: {refval}, crdelt: {refdelta}, refpix: {refpix}')

    for f  in wcsvals.iterrows():
        r = f[1]
        framepath = os.path.join(imdir, r.frameID+'.fits')
        with fits.open(framepath) as hdul:
            framehdr = hdul[0].header
            frameval = np.array([framehdr['CRVAL1'], framehdr['CRVAL2']])
        objpix = np.array([r.CRPIX1, r.CRPIX2])

        deltaval = refval - frameval
        deltapix = deltaval/refdelta
        pred = refpix+deltapix
        
        dist = np.sqrt(((objpix-pred)**2).sum())

        print(f'FrameID: {r.frameID}, Object pix: {objpix}, Predicted: {pred}, distance: {dist:.2f} pixels')
```
